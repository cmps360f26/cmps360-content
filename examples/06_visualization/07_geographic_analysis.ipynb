# 07. Geographic Analysis: Where Does Performance Concentrate?
### Primary Analytical Purpose: *Where does the pattern happen, and does geography explain it?*

Geographic analysis explores how quantitative metrics vary across spatial locations, administrative boundaries, or regional clusters. It addresses three core business questions:
- **Spatial Concentration:** Where are adoption rates or customer activities highest and lowest?
- **Regional Clustering:** Do neighboring territories exhibit similar patterns (spatial autocorrelation)?
- **Resource Allocation:** How should distribution hubs, marketing capital, and physical inventory be deployed geographically?

---

### The Cardinal Rule of Geographic Visualization:
> **Always map a normalized rate (per capita, per 1,000 residents, or per km²), NEVER a raw count.**  
> Mapping raw counts produces an accidental population map rather than an insightful performance visualization.

## 1. Scenario and Dataset Preview
**Scenario:** A regional GCC e-commerce logistics platform evaluates customer purchase intensity across the six Gulf Cooperation Council (GCC) nations: Qatar, Kuwait, the United Arab Emirates, Bahrain, Saudi Arabia, and Oman for 2025.

The management team faces a common decision dilemma:
- **Total volume** (`orders_thousands`) suggests focusing exclusively on Saudi Arabia (182.0k orders).
- **Normalized adoption rate** (`order_rate_per_1000`) reveals that Qatar (7.50) and Kuwait (6.79) lead the region in per-capita digital purchase intensity.

The dataset is loaded from `data/gcc_ecommerce_rates.csv`.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Set modern Seaborn theme
sns.set_theme(style="whitegrid")

# Load GCC geographic dataset
df = pd.read_csv("data/gcc_ecommerce_rates.csv")

# Display the dataset shape and contents
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
df
```

## 2. Primary Visualization: Choropleth Map (Shaded Boundaries)
A **choropleth map** shades administrative regions (countries, provinces, zones) based on a continuous quantitative metric.

### Established Visual Design & Color Best Practices:
1. **Light-to-Dark Sequential Palette:** On light backgrounds, the human eye intuitively associates **darker color intensity with higher magnitude/density** ("more ink = more volume"). We use the sequential `Blues` scale, where darker navy represents high order rates and light blue represents low rates.
2. **Honest Normalization:** Color encodes `order_rate_per_1000`, not raw orders.
3. **Region-Focused Bounding Box:** Center the camera onto the Arabian Gulf with neutral styling for neighboring non-GCC land and ocean.

```python
# --- Plotly Express Implementation: Interactive GCC Choropleth Map ---
# Uses ISO-3 alpha codes to shade country polygons with a sequential light-to-dark palette
fig_map = px.choropleth(
    df, 
    locations="iso_alpha", 
    color="order_rate_per_1000",
    hover_name="country",
    hover_data={
        "iso_alpha": False,
        "population_millions": ":.1f",
        "orders_thousands": ":.1f",
        "order_rate_per_1000": ":.2f"
    },
    color_continuous_scale="Blues",  # Light-to-dark: darker navy = higher rate
    title="<b>GCC E-Commerce: Qatar (7.50) and Kuwait (6.79) Lead in Per-Capita Orders</b>",
    labels={
        "order_rate_per_1000": "Orders / 1k", 
        "population_millions": "Population (M)",
        "orders_thousands": "Total Orders (k)"
    }
)

# Focus viewport onto Arabian Gulf and style background land/ocean cleanly
fig_map.update_geos(
    center=dict(lat=24.5, lon=51.0),
    lataxis_range=[14, 32],
    lonaxis_range=[38, 62],
    visible=False,
    showcountries=True,
    countrycolor="#90A4AE",
    showcoastlines=True,
    coastlinecolor="#78909C",
    showland=True,
    landcolor="#F4F6F9",
    showocean=True,
    oceancolor="#EBF2FA"
)

fig_map.update_layout(
    margin=dict(l=0, r=0, t=50, b=0),
    coloraxis_colorbar=dict(
        title="Orders / 1k",
        ticks="outside",
        thickness=14,
        len=0.75
    )
)

fig_map.show()
```

### Structured Interpretation: The 5-Step Framework
1. **Question:** Which GCC markets demonstrate the highest per-capita digital order intensity?
2. **Visualization:** Regional choropleth map shaded with a sequential light-to-dark `Blues` palette representing orders per 1,000 residents.
3. **Observation:** Qatar leads at 7.50 orders per 1k residents (darkest blue), followed by Kuwait (6.79) and the UAE (6.20). Saudi Arabia has a moderate rate (5.00) despite generating 182k total orders, while Oman has the lowest rate (4.20).
4. **Insight:** High-density, urbanized Gulf city-states have substantially higher digital purchasing frequency than geographically expansive nations with larger rural and dispersed populations.
5. **Implication:** Deploy a dual commercial strategy:
   - **Qatar & UAE:** Focus on premium services, basket frequency, and automated express fulfillment.
   - **Saudi Arabia:** Focus on regional logistics infrastructure and expanding last-mile coverage across secondary cities.

## 3. Alternative 1: Symbol / Bubble Map (Point Coordinates)
While choropleth maps excel at country-wide totals, they suffer from a major cartographic drawback: **the Area-Distortion Dilemma (the "Small Country" Problem)**.
- In a choropleth, large geographic areas (e.g., Saudi Arabia, 2.15M km²) dominate the visual field, while small, high-density territories (Qatar, 11.5k km²; Bahrain, 780 km²) appear physically tiny.
- A **symbol / bubble map** solves this by plotting city hub coordinates where marker **size and color intensity** both encode magnitude, ensuring small nations receive appropriate visual weight.

```python
# --- Plotly Express: Interactive Symbol / Bubble Map ---
# Circle size and dark-intensity both scale with order rate
fig_bubble_map = px.scatter_geo(
    df, 
    lat="latitude", 
    lon="longitude", 
    hover_name="city",
    hover_data={
        "latitude": False,
        "longitude": False,
        "country": True,
        "population_millions": ":.1f",
        "order_rate_per_1000": ":.2f"
    },
    size="order_rate_per_1000", 
    color="order_rate_per_1000",
    color_continuous_scale="Blues",  # Darker blue = higher intensity
    size_max=24,
    title="<b>GCC Capital Hubs: Highest Order Density Concentrated in Gulf Coastal Capitals</b>",
    labels={"order_rate_per_1000": "Orders / 1k"}
)

# Apply identical geographic boundaries and neutral backgrounds
fig_bubble_map.update_geos(
    center=dict(lat=24.5, lon=51.0),
    lataxis_range=[14, 32],
    lonaxis_range=[38, 62],
    visible=False,
    showcountries=True,
    countrycolor="#90A4AE",
    showcoastlines=True,
    coastlinecolor="#78909C",
    showland=True,
    landcolor="#F4F6F9",
    showocean=True,
    oceancolor="#EBF2FA"
)

fig_bubble_map.update_traces(
    marker=dict(line=dict(width=1.2, color="#1A202C"))
)

fig_bubble_map.update_layout(
    margin=dict(l=0, r=0, t=50, b=0),
    coloraxis_colorbar=dict(
        title="Orders / 1k",
        ticks="outside",
        thickness=14,
        len=0.75
    )
)

fig_bubble_map.show()
```

```python
# --- Seaborn / Matplotlib: Schematic Coordinate Scatter ---
# Plots exact coordinates with bubble size and light-to-dark color encoding
plt.figure(figsize=(8.5, 4.8))

ax_scat = sns.scatterplot(
    data=df, 
    x="longitude", 
    y="latitude", 
    size="order_rate_per_1000", 
    hue="order_rate_per_1000", 
    sizes=(150, 600), 
    palette="Blues",  # Darker blue = higher rate
    edgecolor="#1A202C",
    linewidth=1.2,
    legend="brief"
)

# Direct data annotations: city name and rate per 1,000 residents
for _, row in df.iterrows():
    ax_scat.annotate(
        f"{row['city']}\n({row['order_rate_per_1000']:.2f}/1k)", 
        (row["longitude"] + 0.35, row["latitude"] - 0.15), 
        fontsize=9, 
        weight="bold",
        color="#1A202C"
    )

ax_scat.set_xlim(44, 61)
ax_scat.set_ylim(21, 31)
ax_scat.set_title("Schematic Hub Map: City Coordinates Sized & Shaded by Order Rate", fontsize=12, weight="bold", pad=10)
ax_scat.set_xlabel("Longitude (°E)", fontsize=11)
ax_scat.set_ylabel("Latitude (°N)", fontsize=11)

plt.tight_layout()
plt.show()
```

### Interpretation for Symbol Map
- **Observation:** Doha, Kuwait City, Manama, and Abu Dhabi form a clear, high-density cluster along the Arabian Gulf coastline.
- **Derived Insight:** E-commerce delivery operations in the GCC are overwhelmingly driven by coastal metropolitan centers where logistics networks, fulfillment hubs, and consumer purchasing power are concentrated.

## 4. Alternative 2: Sorted Regional Bar Chart (When < 15 Regions)
A crucial tenet of data visualization is knowing **when NOT to use a map**:
- **Decoding Precision:** Human vision decodes bar length aligned on a common baseline with ~99% accuracy, whereas judging subtle 2D color gradations on irregular polygons has much lower accuracy.
- **When N is small (< 15 regions):** A sorted horizontal bar chart communicates rankings and relative gaps instantly, without polygon area bias.
- **Consistent Visual Language:** We use the exact same light-to-dark `Blues` intensity scale so that the visual cue (darker = higher rate) remains consistent across all charts in the analysis.

```python
# Sort descending by order rate
df_sorted = df.sort_values("order_rate_per_1000", ascending=False).reset_index(drop=True)

# Generate light-to-dark colors matching magnitude (darkest for highest)
norm = plt.Normalize(vmin=df_sorted["order_rate_per_1000"].min() * 0.7, vmax=df_sorted["order_rate_per_1000"].max())
bar_colors = [plt.cm.Blues(norm(v)) for v in df_sorted["order_rate_per_1000"]]

# --- Seaborn Implementation: Sorted Bar Chart with Magnitude Shading ---
plt.figure(figsize=(8.5, 4.2))

ax_bar = sns.barplot(
    data=df_sorted, 
    y="country", 
    x="order_rate_per_1000", 
    hue="country",
    palette=bar_colors, 
    legend=False,
    edgecolor="#2D3748",
    linewidth=0.8
)

ax_bar.set_xlim(0, 9)
ax_bar.set_title("When Exact Ranking Matters: Qatar Leads GCC at 7.50 Orders per 1k Residents", fontsize=12, weight="bold", pad=10)
ax_bar.set_xlabel("Orders per 1,000 Residents", fontsize=11)
ax_bar.set_ylabel("", fontsize=11)

# Annotate exact numeric values at the end of each bar
for p in ax_bar.patches:
    val = p.get_width()
    ax_bar.annotate(
        f"{val:.2f}", 
        (val + 0.18, p.get_y() + p.get_height() / 2.), 
        va='center', 
        fontsize=10, 
        fontweight='bold',
        color='#1A202C'
    )

plt.tight_layout()
plt.show()
```

### Interpretation for Regional Bar Chart
- **Observation:** The bar chart removes spatial distraction and makes the ranking unambiguous: Qatar (7.50) leads Kuwait (6.79) by 10.5% and Oman (4.20) by 78.6%.
- **Derived Insight:** When precise comparative gaps matter more than geographical proximity, the sorted bar chart is the most effective and actionable visualization.

## 5. Best Practices & Common Cartographic Pitfalls

| Principle | Best Practice | What to Avoid |
|:---|:---|:---|
| **1. Metric Normalization** | Always map rates per capita, per 1,000 residents, or density. | Never map raw totals on choropleths (creates an accidental population map). |
| **2. Color Encoding** | Use sequential light-to-dark intensity for continuous magnitude on light backgrounds. | Avoid rainbow or inverted palettes where light colors denote high values. |
| **3. Small-Area Visibility** | Use symbol / bubble maps when small territories lead large ones. | Do not let geographically massive regions visually eclipse smaller economic leaders. |
| **4. Map vs. Chart Choice** | Default to a sorted bar chart when N < 15 and ranking is the core message. | Avoid forcing a map when spatial distribution provides no analytical value. |
| **5. Ecological Fallacy** | State regional rates as territory-level aggregates. | Never infer individual consumer behavior solely from aggregate country averages. |

## 6. Key Takeaways
1. **Rates over raw counts:** Choropleth maps must always normalize raw counts by population or geographic area to reflect true intensity.
2. **Darker colors signal higher magnitude:** In sequential choropleths and thematic charts, higher numerical values should consistently map to darker color intensities.
3. **Symbol maps resolve polygon area bias:** Placing sized and shaded bubbles at city coordinates prevents geographically large nations from visually dominating smaller, more productive regions.
4. **Choose bar charts when exact ranking matters:** For fewer than 15 territories, a sorted horizontal bar chart provides faster, more accurate visual decoding than any map.
