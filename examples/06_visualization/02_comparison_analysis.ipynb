# 02. Comparison Analysis: How Do Categories Differ?
### Primary Analytical Purpose: *How do categories differ from one another?*

Comparison analysis evaluates quantitative differences across distinct categories. It addresses questions such as:
- Which category leads, and by how much?
- What is the size of the gap between different channels, products, or departments?
- Does a small group of categories dominate the total?

---

## 1. Scenario and Dataset Preview
**Business Scenario:** A major retail brand operating in Qatar compares annual revenue across five sales channels: *Mobile App*, *Physical Store*, *Online Web*, *Direct Sales*, and *Partner Kiosk*. The data also breaks down revenue by customer segment: *Retail Consumer* vs. *Corporate Client*.

The dataset is loaded from `data/qatar_retail_channels.csv`. Let us load and inspect it.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Set modern Seaborn theme
sns.set_theme(style="whitegrid")

# Load retail channels dataset
df = pd.read_csv("data/qatar_retail_channels.csv")

# Display the dataset
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
df
```

## 2. Primary Visualization: Bar Chart
- **What the chart shows:** Length on a shared zero baseline, representing the total revenue for each channel.
- **When it is useful:** Compare bar lengths from a common zero baseline; longer bars represent higher values.
- ** Use a single color**. Default software palettes assign a rainbow of colours across bars. Change this to use a single color.

```python
# Aggregate total revenue by channel
channel_totals = df.groupby("channel", as_index=False)["revenue_million_qar"].sum()
# Sort descending by value (Principle 6: Sorting is free analysis)
channel_totals = channel_totals.sort_values("revenue_million_qar", ascending=False)

# --- Seaborn Implementation: Static Publication Bar Chart ---
plt.figure(figsize=(8, 4.5))

ax = sns.barplot(
    data=channel_totals, 
    x="channel", 
    y="revenue_million_qar", 
    color="#14A098"  # Principle 7: Single intentional color
)

# Principle 4: Always start at zero — length is the encoding
ax.set_ylim(0, 200)

# Principle 2: Title states the finding; axis carries units
ax.set_title("Mobile App Leads Qatar Retail Sales with 167M QAR", fontsize=13, weight="bold")
ax.set_xlabel("", fontsize=11)
ax.set_ylabel("Revenue (Million QAR)", fontsize=11)

# Annotate each bar with exact value
for p in ax.patches:
    val = p.get_height()
    ax.annotate(f"{val:.0f}M", (p.get_x() + p.get_width() / 2., val + 4), 
                ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# --- Plotly Express Implementation: Interactive Bar Chart ---
fig = px.bar(
    channel_totals, 
    x="channel", 
    y="revenue_million_qar", 
    text_auto=".0f",
    title="Mobile App Leads Qatar Retail Sales with 167M QAR",
    labels={"channel": "", "revenue_million_qar": "Revenue (Million QAR)"}
)

# Single intentional color and honest zero baseline
fig.update_traces(marker_color="#14A098")
fig.update_yaxes(range=[0, 200])
fig.show()
```

### Structured Interpretation: The 5-Step Framework
1. **Question:** Which sales channel generates the highest revenue in Qatar, and how wide is the gap to physical retail?
2. **Visualization:** Bar chart with single color and zero baseline, ordered by revenue.
3. **Observation:** Mobile App generated 167M QAR, outperforming Physical Store (128M QAR) by 30% and Partner Kiosks (30M QAR) by 5.5x.
4. **Insight:** Qatari consumers demonstrate strong mobile-first adoption; digital app transactions now surpass traditional in-store shopping.
5. **Implication:** Prioritize mobile app feature updates (e.g., Apple Pay, Qatar National Address autofill, loyalty integration) over opening new physical kiosks.

## 3. Alternative 1: Grouped Bar Chart (Secondary Split)
- **What the chart shows:** Revenue compared across channels, split by customer segment (*Retail Consumer* vs. *Corporate Client*).
- **When to use instead:** When each category splits by a second categorical variable.
- **Best Practice:** Keep to 2–3 groups per category to prevent visual clutter.

```python
# --- Seaborn: Grouped Bar Chart ---
plt.figure(figsize=(9, 4.5))

ax_grp = sns.barplot(
    data=df, 
    x="channel", 
    y="revenue_million_qar", 
    hue="segment", 
    palette=["#14A098", "#E69F00"]  # Clear, distinct two-segment palette
)

ax_grp.set_ylim(0, 160)
ax_grp.set_title("Direct Sales Relies on Corporate Clients, While Mobile App Dominates Retail", fontsize=12, weight="bold")
ax_grp.set_xlabel("", fontsize=11)
ax_grp.set_ylabel("Revenue (Million QAR)", fontsize=11)
ax_grp.legend(title="Customer Segment")

plt.tight_layout()
plt.show()
```

### Interpretation for Grouped Bar Chart
- **Observation:** Mobile App generates 145M QAR from Retail Consumers and only 22M QAR from Corporate Clients. In contrast, Direct Sales generates 42M QAR from Corporate Clients and only 8M QAR from Retail Consumers.
- **Derived Insight:** Sales channels serve fundamentally distinct audiences: Mobile App and Physical Store are retail-driven, while Direct Sales is an institutional corporate pipeline.

## 5. Best Practices & Common Mistakes
**Visualization Best Practices:**
- **Always start the value axis at zero:** Bar length is the visual encoding. Truncating the axis exaggerates differences.
- **Order bars by value** unless the categories possess a natural sequence (like days of the week).
- **One colour for all bars;** reserve a second colour only to highlight a specific finding.
- **Use horizontal bars when labels are long.**

**Common Mistakes & Misleading Designs:**
- **Rainbow colouring:** Assigning different colors per bar implies groupings or distinctions that do not exist.
- **Truncated axis:** Starting the axis at 100M QAR would make Mobile App look 10x larger than Physical Store instead of 30% larger.
- **3D bars:** Distort perspective and make front bars appear larger than rear bars.

## 6. Key Takeaways
1. **Bar charts are the gold standard for category comparison** because human vision evaluates length on a common baseline with highest precision.
2. **Use a single color:** Rainbow bars create visual noise and false groupings.
3. **Always start the value axis at zero:** Length is the encoding.
4. **Grouped bars work best for 2–3 subcategories** .
